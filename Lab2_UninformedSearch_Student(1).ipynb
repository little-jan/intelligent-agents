## Task

Implement `find_path(start_word, end_word, word_list)` so it returns the
shortest sequence from the start to the end. Consecutive words must
differ in exactly one character. Return `None` when no sequence exists.

For example, one path from `RAM` to `ARM` is
`RAM -> RIM -> AIM -> ARM`.

All words in a returned path must have the same length and occur in the
supplied dictionary. Two dictionaries are provided (one is small, and
the other is large).

**Choose a search strategy before coding.** Each action changes one
letter and has cost 1. Compare the uninformed search strategies covered
in the unit: which are complete for this problem, and which guarantee a
shortest solution under unit action costs? Select and justify a suitable
strategy, then implement it. More than one strategy may satisfy the
requirements under these assumptions.

**Efficiency target:** each supplied case should normally complete in
under one second. Scanning every dictionary word for every expansion is
unnecessarily expensive; consider efficient expansion.

**Checks:** You may test the correctness of your implementation using the
checks provided later.


```python
from collections import deque
def get_neighbours(word, word_list):
    neighbours = []
    alphabet = "ETAOINSHRDLCUMWFGYPBVKJXQZ"

    for i in range(len(word)):
        for letter in alphabet:
            if letter == word[i]:
                continue
            new_word_list = list(word)
            new_word_list[i] = letter
            new_word = "".join(new_word_list)
            if new_word in word_list:
                neighbours.append(new_word)

    return neighbours


# def depth_limited(curr, end, word_list, path, limit):
#     if curr == end:
#         return path
#
#     if limit <= 0:
#         return None
#
#     for neighbour in get_neighbours(curr, word_list):
#         if neighbour not in path:
#             path.append(neighbour)
#             result = depth_limited(neighbour, end, word_list, path, limit-1)
#
#             if result is not None:
#                 return result
#
#             path.pop()
#     return None

def bfs(start_word, end_word, word_list):
    q = deque()
    visited = set()
    visited.add(start_word)

    q.append((start_word, [start_word]))

    path = []
    while q:
        curr, path_so_far = q.popleft()
        path.append(curr)

        if curr == end_word:
            return path_so_far

        for x in get_neighbours(curr, word_list):
            if x not in visited:
                visited.add(x)
                new_path = path_so_far + [x]
                q.append((x, new_path))

    return None

def find_path(start_word, end_word, word_list):
    # start_word and end_word are strings; word_list is a list of strings (i.e., the dictionary).
    # TODO: choose and implement an uninformed search strategy that
    # returns a shortest path as a list of strings from start_word to
    # end_word, or None when no sequence exists.
    word_list = set(word_list)

    for i in range(len(word_list)):
        return bfs(start_word, end_word, word_list)

    return None
```

```python
def validate_word_path(path, start, end, word_list, expected_length=None):
    if expected_length == 0:
        if path is None:
            return True, "correctly returned None"
        return False, f"expected no path, got length {len(path)}"
    if path is None:
        return False, "returned None"
    normalised = [word.upper() for word in path]
    allowed = {word.upper() for word in word_list}
    if not normalised or normalised[0] != start or normalised[-1] != end:
        return False, "incorrect start or end"
    for previous, current in zip(normalised, normalised[1:]):
        if current not in allowed:
            return False, f"{current} is not in the word list"
        if len(current) != len(previous) or sum(a != b for a, b in zip(previous, current)) != 1:
            return False, f"invalid transition: {previous} -> {current}"
    if expected_length is not None and len(normalised) != expected_length:
        return False, f"expected length {expected_length}, got {len(normalised)}"
    return True, f"valid path of length {len(normalised)}"
```

### Small dictionary checks

These cases use the small dictionary defined above.


```python
SMALL_CASES = [
    ("AIM", "BOT", 6),
    ("LOG", "JAM", 5),
    ("LOG", "SUN", 9),
]
for start, end, expected in SMALL_CASES:
    started = time.perf_counter()
    path = find_path(start, end, WORD_LIST_SMALL)
    ok, message = validate_word_path(path, start, end, WORD_LIST_SMALL, expected)
    print(f"{start} -> {end}: {'PASS' if ok else 'NOT YET'} — {message}; "
          f"{time.perf_counter() - started:.4f}s")
    if path is not None:
        print("   ", " -> ".join(path))
```

### Larger dictionary checks

These cases use the embedded dictionary and include different word
lengths as well as an instance with no solution. They provide a
compact correctness and performance check.


```python
LARGE_CHECK_CASES = [
    ("IVY", "EYE", 4),
    ("LYRE", "MAKE", 5),
    ("MINTS", "ROUGE", 9),
    ("BEGIN", "FOUND", 0),
]
for start, end, expected in LARGE_CHECK_CASES:
    started = time.perf_counter()
    path = find_path(start, end, WORD_LIST_LARGE)
    elapsed = time.perf_counter() - started
    ok, message = validate_word_path(
        path, start, end, WORD_LIST_LARGE, expected)
    print(f"{start} -> {end}: {'PASS' if ok else 'NOT YET'} — "
          f"{message}; {elapsed:.4f}s")
```
